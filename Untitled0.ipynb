```python
import numpy as np
import matplotlib.pyplot as ply
from sklearn.datasets import make_blobs
```

```python

n_points = 20
new_classes = 2
```

```python
Data,labels =make_blobs(n_samples=n_points,centers=new_classes,cluster_std=0.5)
ply.scatter(Data[:,0],Data[:,1],c=labels,s=10)
```

```python
def eculidian_dist(A,B):
  xa,ya = A
  xb,yb = B
  dist = np.sqrt((xa-xb)**2+(ya-yb)**2)
  return dist
```

```python
Cc=(5,5)
print(Cc)
Dist=[]
for i in range(n_points):
  A=(Data[i,0],Data[i,1])
  dis =eculidian_dist(A,Cc)
  Dist.append(dis)

print(Dist)


#Data-labels-Dist
```

```python
D=np.column_stack((Data,Dist))
all_data=np.column_stack((Data,labels))
print(all_data)
print(D)
```

```python
sorted_data = all_data[all_data[:,2].argsort()]
print(sorted_data)
```

```python
def majority(sorted_data,K):
  class0=0
  class1=0
  for i in range(K):
    if sorted_data [i,2]== 1:
      class1 +=1
    if sorted_data [i,2]:
      class0 +=1
    if class0>class1:
      return 0
    else:
        return 1
```

```python
print(majority(sorted_data,5))
```
